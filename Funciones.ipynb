## Funciones

Una función es un **subprograma que realiza una tarea específica** y que devuelve  o no resultado(s).

Algunas ventajas de utilizar funciones son:

*  Las funciones son algoritmos más fáciles de diseñar, corregir, mantener y modificar.
*  Evitar la repetición de instrucciones al reutilizarlas mediante una llamada a la función, por lo que el algoritmo  se vuelve más compacto.
*  Las funciones ocultan los detalles de su especificación, de manera que para utilizarlas sólo es necesario conocer su interfaz.

Hasta el momento hemos trabajado con las [funciones integradas](https://docs.python.org/es/3/library/functions.html) (o built-in, en inglés), que responden a esta lógica.

# Funciones propias

Podemos definir nuestras propias funciones. Para ello hacemos uso de la palabra reservada def.

```
def nombre_funcion (<parámetros>):
  <cuerpo>
  return <salidas>
```

La primera línea de la definición de una función la denominaremos, de ahora
en más, encabezado de la función. El encabezado contiene el nombre de la
función y la lista de parámetros.

En Python, una función  puede recibir 0 o más parámetros (expresados entre
entre paréntesis, y separados por comas), efectúa una operación y devuelve
0, 1 o más resultados.

````python
# Por ejemplo

def tiempo_en_segundos(horas, minutos, segundos):
  tiempo = horas * 60 * 60
  tiempo += minutos * 60
  tiempo += segundos
  return tiempo
````

````python
# Invocación

in_tiempo = input("Ingrese un tiempo (HH:MM:SS): ")
hh = int(in_tiempo[:2])
mm = int(in_tiempo[3:5])
ss = float(in_tiempo[6:])


seg = tiempo_en_segundos(hh,mm,ss)
print("Representa ", seg, "segundos.")
````

## Documentación

Es importante documentar las funciones. Es decir, añadir comentarios para indicar cómo deben ser usadas.

Para ello debemos usar la triple comilla `"""` al principio de la función. Se trata de una especie de comentario que podemos usar para indicar cómo la función debe ser usada. No se trata de código, es un simple comentario un tanto especial, conocido como *docstring*.

```
def nombre_funcion (<parámetros>):
  """
  Documentación: que hace, que recibe y que devuelve.
  """
  <cuerpo>
  return <salidas>
```


Ahora cualquier persona que tenga nuestra función, podrá llamar a la función `help()` y obtener la ayuda de cómo debe ser usada.

````python
# Ejemplo
def tiempo_en_segundos_v2(horas, minutos, segundos):
  """
  Función que recibe el tiempo expresado en formato HH:MM:SS y lo devuelve expresado en segundos.
  """

  tiempo = horas * 60 * 60
  tiempo += minutos * 60
  tiempo += segundos
  return tiempo
````

````python
help(tiempo_en_segundos_v2)
````

## Anotaciones

Python es un lenguaje de programación con tipado dinámico, lo que significa que los tipos (int, string, etc) le dan igual.

Sin embargo, en ciertas ocasiones esto nos puede traer problemas. ¿Y si queremos que la función sólo acepte que un argumento sea de un tipo concreto? Pues bien, las anotaciones en funciones o *function annotations* nos permiten especificar los tipos que se esperan recibir.


````python
# Por ejemplo, sin anotaciones:
def duplicar(x):
  return 2*x

doble = duplicar(10)
print(doble)

doble = duplicar ("diez")
print(doble)
````

## Argumentos

Python permite pasar los argumentos de una función de múltiples maneras.

### Por posición

Los argumentos por posición o posicionales son la forma más básica e intuitiva de pasar parámetros. Existirá una correspondencia en el orden entre los parámetros y argumentos de la función al momento de declararse y de invocarse, respectivamente.

La cantidad de parámetros y argumentos debe coincidir, de lo contrario se obtendrá un error.


````python
def resta (a, b):
  return a - b

resta(5,3)
````
